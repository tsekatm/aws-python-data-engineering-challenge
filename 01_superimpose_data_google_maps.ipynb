```
import pandas as pd
import folium
from folium.plugins import MarkerCluster, HeatMap
```

## 🧾 Dataset Overview
We use a CSV that contains geographic coordinates (latitude, longitude), region names, and metric values.  
The goal: visualize each location as a map marker or circle on a base map of South Africa.


```
# Load your dataset (replace with your actual path or URL)
df = pd.read_csv("https://raw.githubusercontent.com/tsekatm/aws-python-data-engineering-challenge/main/sample_data.csv")
df.head()
```

```
# Create a folium map centered on South Africa
m = folium.Map(location=[-29.0, 24.0], zoom_start=6)
marker_cluster = MarkerCluster().add_to(m)

# Add circle markers based on metric values
for _, r in df.iterrows():
    folium.CircleMarker(
        location=[r["latitude"], r["longitude"]],
        radius=r["metric"] / df["metric"].max() * 20 + 3,
        popup=f"{r['region']}: {r['metric']}",
        color="blue", fill=True, fill_color="cyan", fill_opacity=0.6
    ).add_to(marker_cluster)

# Display map
m
```

```
# Optional: Add a heatmap layer for metric intensity
HeatMap(data=df[["latitude", "longitude", "metric"]].values.tolist(),
        radius=15, max_zoom=10).add_to(m)
m
```
